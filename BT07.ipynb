# Bài 4: Vẽ sơ đồ kiến trúc Skip-gram

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

def draw_box(ax, x, y, w, h, text, bg="#EEF2FF", ec="#3B82F6", fontsize=10):
    box = FancyBboxPatch(
        (x - w / 2, y - h / 2), w, h,
        boxstyle="round,pad=0.02,rounding_size=0.02",
        facecolor=bg, edgecolor=ec, linewidth=1.6
    )
    ax.add_patch(box)
    ax.text(x, y, text, ha="center", va="center",
            fontsize=fontsize, fontweight="bold", color="#111827")

def draw_arrow(ax, x1, y1, x2, y2, label="", dx_lbl=0.02, dy_lbl=0.0):
    ax.annotate(
        "", xy=(x2, y2), xytext=(x1, y1),
        arrowprops=dict(arrowstyle="->", lw=1.6, color="#374151")
    )
    if label:
        mx, my = (x1 + x2) / 2 + dx_lbl, (y1 + y2) / 2 + dy_lbl
        ax.text(mx, my, label, fontsize=9, color="#1F2937", va="center")

fig, ax = plt.subplots(figsize=(8, 6.5))
fig.patch.set_facecolor("white")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")

ax.set_title(
    "Kiến trúc Skip-gram (Target: 'tôi', k=1)\nDự đoán 'ngày_mai' và 'không'",
    fontsize=12.5, fontweight="bold", pad=12
)

draw_box(ax, 0.5, 0.85, 0.52, 0.11, "Input: One-hot 'tôi' (8 chiều)", bg="#DBEAFE", ec="#2563EB")
draw_box(ax, 0.5, 0.55, 0.52, 0.12, "Hidden Layer: Vector h (3 chiều)", bg="#FEF3C7", ec="#D97706")
draw_box(ax, 0.24, 0.20, 0.36, 0.13, "Softmax Output 1\nDự đoán 'ngày_mai' (8)", bg="#DCFCE7", ec="#16A34A", fontsize=9.2)
draw_box(ax, 0.76, 0.20, 0.36, 0.13, "Softmax Output 2\nDự đoán 'không' (8)", bg="#DCFCE7", ec="#16A34A", fontsize=9.2)

draw_arrow(ax, 0.5, 0.795, 0.5, 0.61, "W_in (8x3)", dx_lbl=0.03)
draw_arrow(ax, 0.42, 0.49, 0.26, 0.265, "W_out (3x8)", dx_lbl=-0.16, dy_lbl=0.02)
draw_arrow(ax, 0.58, 0.49, 0.74, 0.265, "W_out (3x8)", dx_lbl=0.03, dy_lbl=0.02)

plt.tight_layout()
plt.savefig("skipgram_architecture.png", dpi=300, bbox_inches="tight")
plt.show()
```

# Bài 5: Vẽ sơ đồ kiến trúc CBOW

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

def draw_box(ax, x, y, w, h, text, bg="#EEF2FF", ec="#3B82F6", fontsize=10):
    box = FancyBboxPatch(
        (x - w / 2, y - h / 2), w, h,
        boxstyle="round,pad=0.02,rounding_size=0.02",
        facecolor=bg, edgecolor=ec, linewidth=1.6
    )
    ax.add_patch(box)
    ax.text(x, y, text, ha="center", va="center",
            fontsize=fontsize, fontweight="bold", color="#111827")

def draw_arrow(ax, x1, y1, x2, y2, label="", dx_lbl=0.02, dy_lbl=0.0):
    ax.annotate(
        "", xy=(x2, y2), xytext=(x1, y1),
        arrowprops=dict(arrowstyle="->", lw=1.6, color="#374151")
    )
    if label:
        mx, my = (x1 + x2) / 2 + dx_lbl, (y1 + y2) / 2 + dy_lbl
        ax.text(mx, my, label, fontsize=9, color="#1F2937", va="center")

fig, ax = plt.subplots(figsize=(8, 7))
fig.patch.set_facecolor("white")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")

ax.set_title(
    "Kiến trúc CBOW (Context: 'tôi', 'lập_trình', k=1)\nDự đoán từ mục tiêu 'học'",
    fontsize=12.5, fontweight="bold", pad=12
)

draw_box(ax, 0.24, 0.86, 0.38, 0.10, "Context 1: 'tôi' (8)", bg="#DBEAFE", ec="#2563EB", fontsize=9.5)
draw_box(ax, 0.76, 0.86, 0.38, 0.10, "Context 2: 'lập_trình' (8)", bg="#DBEAFE", ec="#2563EB", fontsize=9.5)

draw_box(ax, 0.24, 0.62, 0.38, 0.09, "Embedding v('tôi') (3)", bg="#E0E7FF", ec="#4F46E5", fontsize=9)
draw_box(ax, 0.76, 0.62, 0.38, 0.09, "Embedding v('lập_trình') (3)", bg="#E0E7FF", ec="#4F46E5", fontsize=9)

draw_box(
    ax, 0.5, 0.40, 0.65, 0.10,
    "Mean Pooling: h = 0.5 * (v_tôi + v_lập_trình) (3)",
    bg="#FEF3C7", ec="#D97706", fontsize=9
)
draw_box(ax, 0.5, 0.16, 0.54, 0.11, "Softmax Output: Dự đoán 'học' (8)", bg="#DCFCE7", ec="#16A34A", fontsize=9.5)

draw_arrow(ax, 0.24, 0.81, 0.24, 0.665, "W_in (8x3)", dx_lbl=0.02)
draw_arrow(ax, 0.76, 0.81, 0.76, 0.665, "W_in (8x3)", dx_lbl=0.02)
draw_arrow(ax, 0.26, 0.575, 0.42, 0.45)
draw_arrow(ax, 0.74, 0.575, 0.58, 0.45)
draw_arrow(ax, 0.5, 0.35, 0.5, 0.215, "W_out (3x8)", dx_lbl=0.03)

plt.tight_layout()
plt.savefig("cbow_architecture.png", dpi=300, bbox_inches="tight")
plt.show()
```
